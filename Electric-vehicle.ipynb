```python
import pandas as pd
import numpy as np

# 1. Load the test set
test_df = pd.read_csv('/content/test.csv')

# Keep track of the 'id' column for the final submission
submission = pd.DataFrame({'id': test_df['id']})

# 2. Replicate all preprocessing steps applied to the training data
# Encode Gender
test_df = pd.get_dummies(test_df, columns=['Gender'], drop_first=True)
test_df['Gender_Male'] = test_df['Gender_Male'].astype(int)
test_df['Gender_Other'] = test_df['Gender_Other'].astype(int)

# Map Home_Charging_Possible and Subsidy_Available to binary integers
test_df.loc[test_df['Home_Charging_Possible'] == 'Yes', 'Home_Charging_Possible'] = 1
test_df.loc[test_df['Home_Charging_Possible'] == 'No', 'Home_Charging_Possible'] = 0
test_df['Home_Charging_Possible'] = test_df['Home_Charging_Possible'].astype(int)

test_df.loc[test_df['Subsidy_Available'] == 'Yes', 'Subsidy_Available'] = 1
test_df.loc[test_df['Subsidy_Available'] == 'No', 'Subsidy_Available'] = 0
test_df['Subsidy_Available'] = test_df['Subsidy_Available'].astype(int)

# Map Range_Anxiety_Level
test_df.loc[test_df['Range_Anxiety_Level'] == 'Low', 'Range_Anxiety_Level'] = 0
test_df.loc[test_df['Range_Anxiety_Level'] == 'Medium', 'Range_Anxiety_Level'] = 1
test_df.loc[test_df['Range_Anxiety_Level'] == 'High', 'Range_Anxiety_Level'] = 2
test_df['Range_Anxiety_Level'] = test_df['Range_Anxiety_Level'].astype(int)

# Encode categorical variables
test_df = pd.get_dummies(test_df, columns=['Current_Car_Type'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['City_Type'], drop_first=True)

test_df['Current_Car_Type_SUV'] = test_df['Current_Car_Type_SUV'].astype(int)
test_df['Current_Car_Type_Sedan'] = test_df['Current_Car_Type_Sedan'].astype(int)
test_df['Current_Car_Type_Truck'] = test_df['Current_Car_Type_Truck'].astype(int)
test_df['City_Type_Suburban'] = test_df['City_Type_Suburban'].astype(int)
test_df['City_Type_Urban'] = test_df['City_Type_Urban'].astype(int)

# 3. Apply the fitted scaler to the numerical features (bypassing name check with .values)
test_df['Age'] = scaler.transform(test_df[['Age']].values)
test_df['Annual_Income_USD'] = scaler.transform(test_df[['Annual_Income_USD']].values)
test_df['Daily_Commute_km'] = scaler.transform(test_df[['Daily_Commute_km']].values)
test_df['Number_of_Cars_Owned'] = scaler.transform(test_df[['Number_of_Cars_Owned']].values)
test_df['Charging_Stations_Near_Home'] = scaler.transform(test_df[['Charging_Stations_Near_Home']].values)
test_df['Charging_Stations_Near_Work'] = scaler.transform(test_df[['Charging_Stations_Near_Work']].values)

# Align columns with training features (drop ID and target variable if present)
x_test_final = test_df[x.columns]

# 4. Generate predictions using the best XGBoost model - now predicting probabilities
test_preds_proba = xgb_model_best.predict_proba(x_test_final)[:, 1] # Get probability of the positive class (1)

# Assign probabilities to the 'Will_Buy_EV' column
submission['Will_Buy_EV'] = test_preds_proba

# 5. Save the final submission
submission.to_csv('submission.csv', index=False)
print("Submission successfully saved to submission.csv")
display(submission.head())
```
